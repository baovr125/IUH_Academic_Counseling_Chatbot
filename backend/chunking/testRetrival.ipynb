```python
# -*- coding: utf-8 -*-
"""
HỆ THỐNG BENCHMARK RETRIEVAL KHÓA LUẬN TỐT NGHIỆP (DÙNG DATA THẬT TỪ CHROMADB)
Các phương pháp: Sparse (BM25), Dense (Vector), Hybrid (RRF)
"""

import os
import time
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from rank_bm25 import BM25Okapi
from sentence_transformers import SentenceTransformer
import chromadb

# ---------------------------------------------------------
# 1. CẤU HÌNH & XỬ LÝ TEXT
# ---------------------------------------------------------
try:
    from underthesea import word_tokenize
    HAS_UNDERTHESEA = True
except ImportError:
    HAS_UNDERTHESEA = False
    

def tokenize(text):
    """Tách từ tiếng Việt, gộp từ ghép nếu có underthesea"""
    if HAS_UNDERTHESEA:
        return word_tokenize(text.lower(), format="text").split()
    return text.lower().split()

# ---------------------------------------------------------
# 2. KHỞI TẠO MÔ HÌNH VÀ KẾT NỐI CHROMADB (DATA THẬT)
# ---------------------------------------------------------
print("📦 Đang khởi tạo mô hình Embedding...")
MODEL_NAME = "sentence-transformers/paraphrase-multilingual-MiniLM-L12-v2"
embedding_model = SentenceTransformer(MODEL_NAME)

CHROMA_DB_PATH = r"G:\Khoa_Luan\Source_code\backend\chroma_db"
COLLECTION_NAME = "camnang_iuh_collection"

print(f"🗄️ Đang kết nối ChromaDB tại: {CHROMA_DB_PATH}")
try:
    chroma_client = chromadb.PersistentClient(path=CHROMA_DB_PATH)
    collection = chroma_client.get_collection(name=COLLECTION_NAME)
    
    # Chỉ cần lấy 'documents' để build BM25, không cần lấy 'embeddings' về RAM nữa
    all_data = collection.get(include=['documents'])
    all_ids = all_data['ids']
    all_docs = all_data['documents']
        
    print(f"✅ Kết nối ChromaDB thành công. Đã tải {len(all_docs)} văn bản để build BM25.")
    
    if len(all_docs) == 0:
        print("⚠️ Cảnh báo: Collection trống. Vui lòng kiểm tra lại dữ liệu.")
        exit()

except Exception as e:
    print(f"❌ Lỗi kết nối ChromaDB: {e}")
    exit()

id_to_doc = dict(zip(all_ids, all_docs))

# Khởi tạo BM25
print("📝 Đang lập chỉ mục BM25...")
tokenized_corpus = [tokenize(doc) for doc in all_docs]
bm25 = BM25Okapi(tokenized_corpus)

# ---------------------------------------------------------
# 3. CÁC HÀM TRUY XUẤT (RETRIEVERS)
# ---------------------------------------------------------
def sparse_retrieval(query, k=5):
    """Truy xuất bằng từ khóa (BM25)"""
    tokenized_query = tokenize(query)
    doc_scores = bm25.get_scores(tokenized_query)
    top_k = sorted(range(len(doc_scores)), key=lambda i: doc_scores[i], reverse=True)[:k]
    return [all_ids[i] for i in top_k]

def dense_retrieval(query, k=5):
    """Truy xuất bằng Vector: Tận dụng trực tiếp sức mạnh tìm kiếm của ChromaDB"""
    query_emb = embedding_model.encode(query).tolist()
    
    # Để ChromaDB tự lo việc tính khoảng cách và trả về Top K
    results = collection.query(
        query_embeddings=[query_emb],
        n_results=k
    )
    
    return results['ids'][0] if results['ids'] else []

def hybrid_rrf_retrieval(query, k=5, rrf_k=60, k_search=30):
    """Truy xuất kết hợp (RRF)"""
    sparse_results = sparse_retrieval(query, k=k_search)
    dense_results = dense_retrieval(query, k=k_search)
    
    rrf_scores = {}
    for rank, chunk_id in enumerate(dense_results, start=1):
        rrf_scores[chunk_id] = rrf_scores.get(chunk_id, 0.0) + (1.0 / (rrf_k + rank))
    for rank, chunk_id in enumerate(sparse_results, start=1):
        rrf_scores[chunk_id] = rrf_scores.get(chunk_id, 0.0) + (1.0 / (rrf_k + rank))
        
    sorted_hybrid = sorted(rrf_scores.items(), key=lambda x: x[1], reverse=True)
    return [item[0] for item in sorted_hybrid[:k]]

# ---------------------------------------------------------
# 4. CHỈ SỐ ĐÁNH GIÁ (METRICS)
# ---------------------------------------------------------
def calculate_hit_rate(retrieved, ground_truth):
    return 1 if ground_truth in retrieved else 0

def calculate_mrr(retrieved, ground_truth):
    for rank, chunk_id in enumerate(retrieved, start=1):
        if chunk_id == ground_truth:
            return 1.0 / rank
    return 0.0

# ---------------------------------------------------------
# 5. PIPELINE KIỂM THỬ VÀ VẼ ĐỒ THỊ
# ---------------------------------------------------------
def visualize_metrics(summary_df):
    sns.set_theme(style="whitegrid")
    
    # Vẽ Hit Rate
    df_hit = summary_df[['Tier', 'Hit Rate (BM25)', 'Hit Rate (Vector)', 'Hit Rate (Hybrid)']].melt(
        id_vars='Tier', var_name='Phương pháp', value_name='Điểm'
    )
    plt.figure(figsize=(9, 5))
    ax = sns.barplot(data=df_hit, x='Tier', y='Điểm', hue='Phương pháp', palette='viridis')
    plt.title('So Sánh Hit Rate @5 Theo Tier')
    plt.ylim(0, 1.15)
    for c in ax.containers: ax.bar_label(c, fmt='%.2f')
    plt.tight_layout()
    plt.savefig('Hit_Rate_Chart.png', dpi=300)
    
    # Vẽ MRR
    df_mrr = summary_df[['Tier', 'MRR (BM25)', 'MRR (Vector)', 'MRR (Hybrid)']].melt(
        id_vars='Tier', var_name='Phương pháp', value_name='Điểm'
    )
    plt.figure(figsize=(9, 5))
    ax2 = sns.barplot(data=df_mrr, x='Tier', y='Điểm', hue='Phương pháp', palette='magma')
    plt.title('So Sánh MRR @5 Theo Tier')
    plt.ylim(0, 1.15)
    for c in ax2.containers: ax2.bar_label(c, fmt='%.2f')
    plt.tight_layout()
    plt.savefig('MRR_Chart.png', dpi=300)
    plt.close('all')
    print("\n📈 Đã lưu đồ thị: 'Hit_Rate_Chart.png' và 'MRR_Chart.png'")

def run_benchmark(test_dataset, k_value=5, output_file="Retrieval_Benchmark.xlsx"):
    print(f"\n🚀 Chạy Benchmark với {len(test_dataset)} câu hỏi (Top K={k_value})...")
    results = []
    
    for item in test_dataset:
        query = item['query']
        gt_id = item['ground_truth_chunk_id']
        tier = f"Tier {item['tier']}"
        
        t0 = time.time()
        sparse_res = sparse_retrieval(query, k=k_value)
        lat_sparse = (time.time() - t0) * 1000
        
        t0 = time.time()
        dense_res = dense_retrieval(query, k=k_value)
        lat_dense = (time.time() - t0) * 1000
        
        t0 = time.time()
        hybrid_res = hybrid_rrf_retrieval(query, k=k_value)
        lat_hybrid = (time.time() - t0) * 1000
        
        results.append({
            "Tier": tier,
            "Câu hỏi": query,
            "ID Đáp án": gt_id,
            "Hit Rate (BM25)": calculate_hit_rate(sparse_res, gt_id),
            "Hit Rate (Vector)": calculate_hit_rate(dense_res, gt_id),
            "Hit Rate (Hybrid)": calculate_hit_rate(hybrid_res, gt_id),
            "MRR (BM25)": round(calculate_mrr(sparse_res, gt_id), 4),
            "MRR (Vector)": round(calculate_mrr(dense_res, gt_id), 4),
            "MRR (Hybrid)": round(calculate_mrr(hybrid_res, gt_id), 4),
            "Latency BM25 (ms)": round(lat_sparse, 2),
            "Latency Vector (ms)": round(lat_dense, 2),
            "Latency Hybrid (ms)": round(lat_hybrid, 2)
        })
        
    df = pd.DataFrame(results)
    
    numeric_cols = [c for c in df.columns if "Hit" in c or "MRR" in c or "Latency" in c]
    summary_df = df.groupby('Tier')[numeric_cols].mean().reset_index()
    
    visualize_metrics(summary_df)
    
    with pd.ExcelWriter(output_file, engine='openpyxl') as writer:
        summary_df.to_excel(writer, sheet_name='Tong_Ket', index=False)
        df.to_excel(writer, sheet_name='Chi_Tiet', index=False)
        
        for sheet in writer.book.sheetnames:
            ws = writer.book[sheet]
            for col in ws.columns:
                ws.column_dimensions[col[0].column_letter].width = 20

    print(f"✅ Báo cáo Excel đã lưu tại: {output_file}\n")
    

def print_top5_analysis_per_tier(test_dataset, k=5):
    """
    Phân tích trực quan: In ra Top 5 kết quả của 3 phương pháp cho mỗi Tier.
    Giúp hội đồng thấy rõ sự khác biệt về chất lượng truy xuất.
    """
    print(f"\n" + "="*80)
    print(f"📊 PHÂN TÍCH TRỰC QUAN (VISUALIZATION TEXT) - TOP {k} KẾT QUẢ THEO TIER")
    print("="*80)
    
    # Nhóm các câu hỏi theo Tier
    tiers = sorted(list(set([item['tier'] for item in test_dataset])))
    
    for tier in tiers:
        # Tìm câu hỏi đầu tiên hợp lệ (có đáp án) của Tier này để làm mẫu
        sample_item = next((item for item in test_dataset if item['tier'] == tier and item['ground_truth_chunk_id'] != 'NOT_FOUND_THIEU_DATA'), None)
        
        if not sample_item:
            continue
            
        query = sample_item['query']
        gt_id = sample_item['ground_truth_chunk_id']
        
        print(f"\n" + "⭐"*40)
        print(f"📌 TIER {tier} - CÂU HỎI ĐẠI DIỆN: '{query}'")
        print(f"🎯 ĐÁP ÁN ĐÚNG (GROUND TRUTH ID): {gt_id}")
        print("⭐"*40)
        
        # Chạy truy xuất Top 5
        sparse_res = sparse_retrieval(query, k=k)
        dense_res = dense_retrieval(query, k=k)
        hybrid_res = hybrid_rrf_retrieval(query, k=k)
        
        methods = [
            ("BM25 (SPARSE)", sparse_res),
            ("VECTOR (DENSE)", dense_res),
            ("HYBRID (RRF)", hybrid_res)
        ]
        
        # In kết quả cho từng phương pháp
        for method_name, results in methods:
            print(f"\n 🟢 PHƯƠNG PHÁP: {method_name}")
            print(f" --------------------------------------------------")
            for rank, chunk_id in enumerate(results, start=1):
                # Đánh dấu highlight nếu chunk này chính là đáp án đúng
                hit_marker = "✅ [HIT]" if chunk_id == gt_id else "❌ [MISS]"
                
                # Lấy nội dung text (cắt ngắn 100 ký tự để dễ nhìn)
                text_content = id_to_doc.get(chunk_id, 'Không tìm thấy nội dung văn bản.')
                short_text = text_content[:120].replace('\n', ' ') + "..." if len(text_content) > 120 else text_content.replace('\n', ' ')
                
                print(f"   {rank}. {hit_marker} ID: {chunk_id}")
                print(f"      📝 Text: {short_text}")
# ---------------------------------------------------------
# 6. KHỞI CHẠY (MAIN)
# ---------------------------------------------------------
if __name__ == "__main__":
    
    real_test_data = [
        # TIER 1
    {'query': 'Điểm VSTEP tối thiểu để đạt Bậc 3 là bao nhiêu?', 'ground_truth_chunk_id': '6754b800-d1ae-49cc-8387-d88c880a890c', 'tier': 1},
    {'query': 'Mức điểm Viết của chứng chỉ TOEIC 4 kỹ năng đối với Bậc 4 là bao nhiêu?', 'ground_truth_chunk_id': '6754b800-d1ae-49cc-8387-d88c880a890c', 'tier': 1},
    {'query': 'Ai đang giữ chức vụ Bí thư Đảng ủy, Hiệu trưởng Đại học Công nghiệp TP.HCM?', 'ground_truth_chunk_id': '77fd6196-7356-41d8-822a-d9b1b9d9ec9c', 'tier': 1},
    {'query': 'Trường Đại học Công nghiệp Thành phố Hồ Chí Minh có truyền thống đào tạo bao nhiêu năm?', 'ground_truth_chunk_id': 'NOT_FOUND_THIEU_DATA', 'tier': 1}, 
    {'query': 'Nhà trường nhận Huân chương Độc lập Hạng 3 vào năm nào?', 'ground_truth_chunk_id': 'NOT_FOUND_THIEU_DATA', 'tier': 1},
    {'query': 'Nhà trường nhận Huân chương Lao động Hạng 1 vào năm nào?', 'ground_truth_chunk_id': 'NOT_FOUND_THIEU_DATA', 'tier': 1},
    {'query': 'Lịch khám sức khỏe cho sinh viên K19 và K20 được thông báo vào ngày nào?', 'ground_truth_chunk_id': 'NOT_FOUND_THIEU_DATA', 'tier': 1},
    {'query': 'Nếu cần hỗ trợ kỹ thuật, sinh viên có thể gọi số điện thoại nào?', 'ground_truth_chunk_id': 'b78da5dc-57ea-453a-8501-195d9d5c2533', 'tier': 1}, # Có nhắc đến hỗ trợ kỹ thuật
    {'query': 'Phòng Công tác Chính trị và Hỗ trợ Sinh viên có thể được liên hệ qua số máy nhánh nào?', 'ground_truth_chunk_id': '2f60b360-629f-4c6c-9b49-90758f90fcfe', 'tier': 1}, # Nhắc đến Phòng CTCT
    {'query': 'Website Cẩm nang người học do đơn vị nào phát triển?', 'ground_truth_chunk_id': 'NOT_FOUND_THIEU_DATA', 'tier': 1},
    
    # TIER 2
    {'query': 'Trường có chấp nhận bằng TOEFL iBT phiên bản làm tại nhà không?', 'ground_truth_chunk_id': 'NOT_FOUND_THIEU_DATA', 'tier': 2},
    {'query': 'Sinh viên muốn nộp chứng chỉ tiếng Anh thi bên ngoài thì nên thi ở những chỗ nào để được công nhận?', 'ground_truth_chunk_id': 'd20bdaf1-ee0e-4c92-9989-410e7fdc8186', 'tier': 2}, # Danh mục chứng chỉ ngoại ngữ
    {'query': 'Chuyên trang Cẩm nang người học có bao nhiêu nhóm thông tin chính để sinh viên tra cứu?', 'ground_truth_chunk_id': 'NOT_FOUND_THIEU_DATA', 'tier': 2},
    {'query': 'Chỗ nào cung cấp thông tin về các hoạt động giao lưu quốc tế của trường?', 'ground_truth_chunk_id': '33e3e9b8-dc92-4941-9c25-efb77d2c9629', 'tier': 2},
    {'query': 'Thông báo về việc đóng bảo hiểm y tế cuối năm 2025 cho sinh viên khóa mới nhất được đăng khi nào?', 'ground_truth_chunk_id': 'fbae1105-243e-4c6a-8a7a-d0630a4d7e62', 'tier': 2}, # Có chunk về BHYT
    {'query': 'Trang web nào để sinh viên xem thông tin về Ký túc xá?', 'ground_truth_chunk_id': '4b3c6c71-de93-4935-9aca-e2936c59323b', 'tier': 2},
    {'query': 'Sự kiện chào đón tân sinh viên năm 2025 được nhà trường ra thông báo vào thời gian nào?', 'ground_truth_chunk_id': 'f60c4002-17b0-4182-8ae1-0dfb35381b3b', 'tier': 2},
    {'query': 'Thư chúc mừng tân sinh viên đầu năm học là do ai viết?', 'ground_truth_chunk_id': 'd5ddf6a3-ac7f-4b05-abc9-2144ee1fd93b', 'tier': 2},
    {'query': 'Sinh viên có thể tìm mã khoa của mình ở đâu?', 'ground_truth_chunk_id': 'a00cfd41-5a06-41d6-b42f-7a9f447e4d4c', 'tier': 2},
    {'query': 'Sinh viên ngành Kế toán, Kiểm toán thuộc viện nào quản lý?', 'ground_truth_chunk_id': '3f76f009-4910-4411-a9f8-8b4980b9b97d', 'tier': 2},
    
    # TIER 3 (Những câu phức tạp có thể tìm thấy phần nào thông tin)
    {'query': 'Điểm TOEFL iBT 94 tương đương với khung CEFR nào và mức VSTEP tương ứng là bao nhiêu?', 'ground_truth_chunk_id': '6754b800-d1ae-49cc-8387-d88c880a890c', 'tier': 3},
    {'query': 'Hình ảnh nhập học của Tân sinh viên K21 được chụp liên tục trong bao nhiêu ngày kể từ 23/08/2025?', 'ground_truth_chunk_id': 'NOT_FOUND_THIEU_DATA', 'tier': 3},
    {'query': 'Ngành Quản trị kinh doanh và ngành Marketing của Khoa Quản trị kinh doanh có sinh hoạt đầu khóa cùng một ngày không?', 'ground_truth_chunk_id': '3f76f009-4910-4411-a9f8-8b4980b9b97d', 'tier': 3},
    {'query': 'Khoa Công nghệ Điện tử và Khoa Công nghệ Hóa học có tổ chức sinh hoạt đầu khóa vào cùng một ngày không?', 'ground_truth_chunk_id': '299155ce-7c1e-4da0-9e2b-596308c37ac0', 'tier': 3},
    {'query': 'Khoa Ngoại ngữ và Khoa Công nghệ May - Thời trang, khoa nào tổ chức sinh hoạt đầu khóa sớm hơn?', 'ground_truth_chunk_id': '299155ce-7c1e-4da0-9e2b-596308c37ac0', 'tier': 3},
    {'query': 'Sinh hoạt đầu khoá của Khoa Công nghệ Cơ khí và Khoa Công nghệ Động lực diễn ra vào ngày nào?', 'ground_truth_chunk_id': '299155ce-7c1e-4da0-9e2b-596308c37ac0', 'tier': 3},
    {'query': 'Khoa Công nghệ Điện và Khoa Công nghệ Thông tin tổ chức sinh hoạt vào ngày nào?', 'ground_truth_chunk_id': '299155ce-7c1e-4da0-9e2b-596308c37ac0', 'tier': 3},
    {'query': 'Viện Tài chính - Kế toán tổ chức sinh hoạt cho ngành Kế toán, Kiểm toán và ngành Tài chính - Ngân hàng có cùng thời gian không?', 'ground_truth_chunk_id': '299155ce-7c1e-4da0-9e2b-596308c37ac0', 'tier': 3},
    {'query': 'Khoa Luật và Khoa học chính trị sinh hoạt đầu khóa trước hay sau ngày ra thông báo Ngày hội Chào Tân sinh viên 2025?', 'ground_truth_chunk_id': 'f60c4002-17b0-4182-8ae1-0dfb35381b3b', 'tier': 3},
    {'query': 'Khoa Thương mại và Du lịch sinh hoạt đầu khóa vào ngày nào và thông báo khám sức khỏe K19, K20 có trước đó mấy ngày?', 'ground_truth_chunk_id': 'NOT_FOUND_THIEU_DATA', 'tier': 3},
]
    
    

    print_top5_analysis_per_tier(real_test_data, k=5)
    run_benchmark(real_test_data, k_value=5)
```
