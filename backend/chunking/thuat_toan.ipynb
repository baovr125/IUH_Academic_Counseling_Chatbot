```python
# Cell 1: Khởi tạo, thiết lập dữ liệu và thông số BM25
import math
import numpy as np

print("--- THÔNG TIN DỮ LIỆU ĐẦU VÀO ---")
query = "buộc thôi học thi hộ"
keywords = ["buộc", "thôi", "học", "thi", "hộ"]
print(f"Truy vấn (Q): '{query}'[cite: 1]")
print(f"Từ khóa (9): {keywords}[cite: 1]")

chunks = {
    "Chunk 1": "Sau mỗi học kỳ, Nhà trường sẽ xem xét và ra quyết định buộc thôi học nếu sinh viên thuộc một trong các trường hợp sau",
    "Chunk 2": "a) Có số lần cảnh báo kết quả học tập vượt quá 2 lần liên tiếp",
    "Chunk 3": "c) Bị kỷ luật lần thứ hai vì lý do thi hộ hoặc nhờ người thì hộ trong các kỳ thi hoặc bị kỷ luật ở mức xoá tên khỏi danh sách sinh viên của Trường"
}

print("\nTập dữ liệu (Chunks):")
for key, text in chunks.items():
    print(f"- {key}: '{text}'[cite: 1]")

# Thông số toán học
k1 = 1.5 # Hệ số bão hòa TF[cite: 1]
b = 0.75 # Hệ số phạt chiều dài[cite: 1]
avgdl = 26 # Độ dài trung bình của các chunk[cite: 1]

def mock_idf(word):
    return 0.98

def calculate_bm25_term(tf, dl, idf):
    numerator = tf * (k1 + 1)
    denominator = tf + k1 * (1 - b + b * (dl / avgdl))
    return idf * (numerator / denominator)
```

```python
# Cell 2: Demo BM25 cho Chunk 3
print("\n--- GIẢI PHẪU ĐIỂM BM25 CHO CHUNK 3 ---")
print(f"Tập trung vào tính điểm của từ khóa 'thi' trong Chunk 3[cite: 1]")
print(f"Nội dung: '{chunks['Chunk 3']}'")

# Thông số của Chunk 3
dl_chunk3 = 36 # Độ dài tài liệu[cite: 1]
tf_thi = 3     # Tần suất xuất hiện của từ "thi"[cite: 1]
idf_thi = mock_idf("thi") 

print(f"\nThông số tính toán: Chiều dài (dl) = {dl_chunk3}, Tần suất (TF) = {tf_thi}, Độ hiếm giả định (IDF) = {idf_thi}")

# 1. Tử số (TF bão hòa)
numerator = tf_thi * (k1 + 1)
print(f"1. Tử số (TF bão hòa): {numerator}[cite: 1]")

# 2. Hình phạt chiều dài
length_penalty = 1 - b + b * (dl_chunk3 / avgdl)
print(f"2. Hình phạt chiều dài: {length_penalty:.3f}[cite: 1]")

# 3. Mẫu số
denominator = tf_thi + k1 * length_penalty
print(f"3. Mẫu số: {denominator:.3f}[cite: 1]")

# 4. Điểm số đóng góp
score_thi = idf_thi * (numerator / denominator)
print(f"=> Điểm BM25 cho từ 'thi' trong Chunk 3: {score_thi:.3f}")

print("\n--- TỔNG KẾT XẾP HẠNG BM25 ---")
print("Hạng 1: Chunk 3 (2.73 điểm)[cite: 1]")
print("Hạng 2: Chunk 1 (2.63 điểm)[cite: 1]")
print("Hạng 3: Chunk 2 (0.56 điểm)[cite: 1]")
```

```python
# Cell 3: Demo Cosine Similarity
print("--- COSINE SIMILARITY ---")

def cosine_similarity(vec1, vec2):
    dot_product = np.dot(vec1, vec2)
    magnitude = np.linalg.norm(vec1) * np.linalg.norm(vec2)
    if magnitude == 0:
        return 0
    return dot_product / magnitude

print("Từ điển: [mèo, ăn, cá, chuột, chó, gặm, xương][cite: 1]")

# Câu văn và vector
sentences = {
    "A": {"text": "Mèo ăn cá", "vector": np.array([1, 1, 1, 0, 0, 0, 0])},
    "B": {"text": "Mèo ăn chuột", "vector": np.array([1, 1, 0, 1, 0, 0, 0])},
    "C": {"text": "Chó gặm xương", "vector": np.array([0, 0, 0, 0, 1, 1, 1])}
}

for key, data in sentences.items():
    print(f"Câu {key}: '{data['text']}' -> Biểu diễn Vector: {data['vector'].tolist()}[cite: 1]")

print("\nKết quả so sánh:")
sim_AB = cosine_similarity(sentences["A"]["vector"], sentences["B"]["vector"])
print(f"So sánh A và B: Cosine = {sim_AB:.3f} -> Ngữ nghĩa tương đồng.[cite: 1]")

sim_AC = cosine_similarity(sentences["A"]["vector"], sentences["C"]["vector"])
print(f"So sánh A và C: Cosine = {sim_AC:.3f} -> Vuông góc, không liên quan.[cite: 1]")
```

```python
# Cell 4: Demo Rank Fusion (RRF)
print("--- RECIPROCAL RANK FUSION (RRF) ---")

query_rrf = "Quy định xử phạt kỷ luật thi giùm người khác"
print(f"Truy vấn dung hợp: '{query_rrf}'[cite: 1]\n")

k = 60 # Hằng số RRF[cite: 1]

# Bảng xếp hạng đầu vào
ranks = {
    "Chunk 1": {"bm25_rank": 2, "vector_rank": 1},
    "Chunk 3": {"bm25_rank": 1, "vector_rank": 3},
    "Chunk 2": {"bm25_rank": 3, "vector_rank": 2},
}

print("Thứ hạng ban đầu:")
for chunk, rank_data in ranks.items():
    print(f"- {chunk}: Hạng BM25 = {rank_data['bm25_rank']}, Hạng Vector = {rank_data['vector_rank']}[cite: 1]")

def calculate_rrf(rank1, rank2, k_val=60):
    return (1 / (k_val + rank1)) + (1 / (k_val + rank2))

print("\nQuá trình tính toán RRF Score (k=60):")
rrf_scores = {}
for chunk, rank_data in ranks.items():
    score = calculate_rrf(rank_data["bm25_rank"], rank_data["vector_rank"], k)
    rrf_scores[chunk] = score
    print(f"{chunk}: 1/({k}+{rank_data['bm25_rank']}) + 1/({k}+{rank_data['vector_rank']}) = {score:.4f}[cite: 1]")

print("\nKết quả xếp hạng RRF cuối cùng:")
sorted_rrf = sorted(rrf_scores.items(), key=lambda item: item[1], reverse=True)
for i, (chunk, score) in enumerate(sorted_rrf, 1):
    print(f"Hạng {i}: {chunk} (Score: {score:.4f})")
```

```python
# Cell 5: Demo Weighted Fusion (Alpha)
print("--- WEIGHTED FUSION (ALPHA) ---")
print("Bài toán: Ưu tiên tìm kiếm ngữ nghĩa với Trọng số 70% Vector Search và 30% BM25 (Alpha = 0.7)[cite: 1]\n")

# Điểm thô giả định
raw_bm25 = {"Chunk 1": 2.63, "Chunk 2": 0.56, "Chunk 3": 2.73}
print("Điểm BM25 thô ban đầu:")
for chunk, score in raw_bm25.items():
    print(f"- {chunk}: {score}[cite: 1]")

def min_max_scaling(x, x_min, x_max):
    return (x - x_min) / (x_max - x_min)

min_bm25 = min(raw_bm25.values())
max_bm25 = max(raw_bm25.values())

print("\nVí dụ Chuẩn hóa Min-Max cho Chunk 1:")
norm_chunk1_bm25 = min_max_scaling(raw_bm25["Chunk 1"], min_bm25, max_bm25)
print(f"X_norm = ({raw_bm25['Chunk 1']} - {min_bm25}) / ({max_bm25} - {min_bm25}) = {norm_chunk1_bm25:.3f}[cite: 1]\n")

# Bảng điểm sau chuẩn hóa
norm_scores = {
    "Chunk 1": {"bm25": 0.954, "vector": 1.000},
    "Chunk 2": {"bm25": 0.000, "vector": 0.468},
    "Chunk 3": {"bm25": 1.000, "vector": 0.000},
}

alpha = 0.7 # Trọng số Vector Search[cite: 1]
final_scores = {}

print("Quá trình tính điểm dung hợp:")
for chunk, scores in norm_scores.items():
    s_vec = scores["vector"]
    s_bm25 = scores["bm25"]
    final_score = (alpha * s_vec) + ((1 - alpha) * s_bm25)
    final_scores[chunk] = final_score
    print(f"{chunk}: (0.7 x {s_vec:.3f}) + (0.3 x {s_bm25:.3f}) = {final_score:.3f}[cite: 1]")

print("\nBảng xếp hạng tổng thể cuối cùng:")
sorted_final = sorted(final_scores.items(), key=lambda item: item[1], reverse=True)
for i, (chunk, score) in enumerate(sorted_final, 1):
    print(f"Hạng {i}: {chunk} ({score:.3f})")
```
